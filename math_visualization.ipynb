```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 200)
y1 = x
y2 = x**2
y3 = np.sin(x)
y4 = np.exp(-0.1 * x) * np.cos(x)

plt.plot(x, y1, label='y = x', linestyle='-', color="blue")
plt.plot(x, y2, label='y = x²', linestyle='--', color="green")
plt.plot(x, y3, label='y = sin(x)', linestyle=':')
plt.plot(x, y4, label='y = e^(-0.1x) cos(x)', linestyle='-.', color="red")

plt.title("Mathematical Function Visualization")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()
plt.grid(True)

plt.savefig("function_plot.png")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 500)

y = np.exp(-0.1 * x) * np.sin(2 * x) + 0.5 * x

plt.plot(x, y)

plt.title("Own Equation")
plt.xlabel("x")
plt.ylabel("y")

plt.grid(True)
plt.savefig("own_equation.png")
plt.show()

# This graph combines exponential decay with a sine wave
```

```python
students = ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"]
midterm = [85, 72, 90, 66, 78, 92, 60, 74, 88, 95]
final = [80, 70, 94, 68, 75, 90, 65, 72, 84, 96]

total = 0.4*np.array(midterm) + 0.6 * np.array(final)

plt.figure(figsize=(6, 4))

plt.scatter(midterm, final)

plt.title("Midterm vs Final Scores")
plt.xlabel("Midterm Score")
plt.ylabel("Final Score")
plt.grid(True)

plt.savefig("score_scatter.png")
plt.show()


plt.figure(figsize=(6, 4))

plt.hist(total, bins=5)

plt.title("Distribution of Total Scores")
plt.xlabel("Total Score")
plt.ylabel("Frequency")
plt.grid(True)

plt.savefig("score_histogram.png")
plt.show()

plt.figure(figsize=(8, 4))

plt.bar(students, total)

plt.title("Student Total Scores")
plt.xlabel("Students")
plt.ylabel("Total Score")
plt.grid(True)

plt.savefig("score_bar_chart.png")
plt.show()
```

```python
slope, intercept = np.polyfit(midterm, final, 1)

x = np.linspace(50, 100, 200)
y = slope * x + intercept

plt.scatter(midterm, final, label="Actual Scores")

plt.plot(x, y, linestyle='--', label="Best-Fit Line")

plt.title("Midterm vs Final Score Prediction")
plt.xlabel("Midterm Score")
plt.ylabel("Final Score")

plt.legend()
plt.grid(True)

plt.savefig("score_prediction.png")

plt.show()

predict_50 = slope * 50 + intercept
predict_75 = slope * 75 + intercept
predict_100 = slope * 100 + intercept

print("Predicted final score for midterm = 50:", predict_50)
print("Predicted final score for midterm = 75:", predict_75)
print("Predicted final score for midterm = 100:", predict_100)
```
